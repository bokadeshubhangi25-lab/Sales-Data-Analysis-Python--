```python
import pandas as pd

data = {
    "Product": ["Laptop","Moblie","Tablet","Laptop","Moblie","Headphones","Tablet","Laptop","Earbuds","Moblie","Headphone"],

    "Category": ["Electronics","Electronics","Electronics","Electronics","Electronics","Accessories","Electronics","Electronics","Electronics","Accessories","Accessories"],

    "City": ["Pune","Mumbai","Akola","Nagpur","Pune","Delhi","Mumbai","Nagpur","Delhi","Pune","Akola" ],

    "Quantity": [3,5,3,2,4,10,6,8,5,7,11],

    "Price": [50000,20000,15000,60000,20000,3000,15000,25000,55000,25000,1800],

    "Profit": [10000,9000,6000,11000,4000,1000,7000,8000,10000,8000,1000],


}

df = pd.DataFrame(data)

df["Sales"] = df["Quantity"] * df["Price"] 

df.to_csv("sales_data.csv",index=False)

print("Dataset Created Successfully!")
print(df)
```

```python
total_sales = df["Sales"].sum()
print("Total Sales:",total_sales)
```

```python
total_profit = df["Profit"].sum()
print("Total Profit:",total_profit)
```

```python
product_sales = df.groupby("Product")["Sales"].sum()
print(product_sales.sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

Product_Sales = df.groupby("Product")["Sales"].sum()

Product_Sales.plot(kind="bar", figsize=(8, 5))

plt.title("Product Wise Sales Analysis")
plt.xlabel("Product")
plt.ylabel("Totale Sales")
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```

```python
City_Sales = df.groupby("City")["Sales"].sum()

City_Sales.plot(kind="bar", figsize=(8, 5))

plt.title("City Wise Sales Analysis")
plt.xlabel("City")
plt.ylabel("Sales Amoutnt")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
Category_Profit = df.groupby("Category")["Profit"].sum()

Category_Profit.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7)
)

plt.title("Category Wise Profit")
plt.ylabel("")
plt.show()
```

```python
print("BUSINESS INSIGHTS")

print("Total Revenue:", df["Sales"].sum())

print("Total Profit:", df["Profit"].sum())

print("Best Sellling Product:",
      df.groupby("Product")["Sales"].sum().idxmax())

print("Best Performing City:",
      df.groupby("City")["Sales"].sum().idxmax())

print("Most Profitable Category:",
      df.groupby("Category")["Profit"].sum().idxmax())
```
